General Revenue Overview
1. Total Revenue
We'll sum up the Total Purchase Amount.

2. Average Order Value (AOV)
This is the average revenue per purchase.

3. Total Orders
This equals the number of rows (transactions).

4. Top 5 Product Categories by Revenue
5. Revenue Trends by Year (based on Purchase Date)

1, Total Revenue

```python
# Convert 'Purchase Date' to datetime
df['Purchase Date'] = pd.to_datetime(df['Purchase Date'])

# 1. Total Revenue
total_revenue = df['Total Purchase Amount'].sum()

# 2. Average Order Value (AOV)
average_order_value = df['Total Purchase Amount'].mean()

# 3. Total Orders
total_orders = df.shape[0]

# 4. Top 5 Product Categories by Revenue
top_categories = df.groupby('Product Category')['Total Purchase Amount'] \
                   .sum().sort_values(ascending=False).head(5)

# 5. Revenue Trends by Year
df['Year'] = df['Purchase Date'].dt.year
revenue_by_year = df.groupby('Year')['Total Purchase Amount'].sum()

# Display Results
print(f"Total Revenue: ${total_revenue:,.2f}")
print(f"Average Order Value (AOV): ${average_order_value:,.2f}")
print(f"Total Orders: {total_orders}")
print("\nTop 5 Product Categories by Revenue:\n", top_categories)
print("\nRevenue by Year:\n", revenue_by_year)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load your dataset

df['Purchase Date'] = pd.to_datetime(df['Purchase Date'])

# ----------------- Data Prep -----------------
df['Year-Month'] = df['Purchase Date'].dt.to_period('M').astype(str)

# Age group bins
bins = [0, 20, 30, 40, 50, 60, 70, 100]
labels = ['<20', '20–29', '30–39', '40–49', '50–59', '60–69', '70+']
df['Age Group'] = pd.cut(df['Customer Age'], bins=bins, labels=labels, right=False)

# ----------------- Analysis -----------------

# 1. Monthly Revenue
monthly_revenue = df.groupby('Year-Month')['Total Purchase Amount'].sum()

# 2. Top 10 Customers by Revenue
top_customers = df.groupby('Customer ID')['Total Purchase Amount'].sum().sort_values(ascending=False).head(10)

# 3. Return Impact
df['Return Value'] = df['Total Purchase Amount'] * df['Returns']
total_return_loss = df['Return Value'].sum()

# 4. Revenue by Gender
revenue_by_gender = df.groupby('Gender')['Total Purchase Amount'].sum()

# 5. Revenue by Age Group
revenue_by_age_group = df.groupby('Age Group')['Total Purchase Amount'].sum()

# ----------------- Visualizations -----------------
sns.set(style="whitegrid")

# Monthly Revenue Trend
plt.figure(figsize=(12, 6))
monthly_revenue.plot(marker='o')
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Top Customers
plt.figure(figsize=(10, 5))
sns.barplot(x=top_customers.index.astype(str), y=top_customers.values, palette="magma")
plt.title("Top 10 Customers by Revenue")
plt.xlabel("Customer ID")
plt.ylabel("Revenue ($)")
plt.tight_layout()
plt.show()

# Revenue by Gender
plt.figure(figsize=(6, 4))
sns.barplot(x=revenue_by_gender.index, y=revenue_by_gender.values, palette="pastel")
plt.title("Revenue by Gender")
plt.xlabel("Gender")
plt.ylabel("Revenue ($)")
plt.tight_layout()
plt.show()

# Revenue by Age Group
plt.figure(figsize=(8, 5))
sns.barplot(x=revenue_by_age_group.index, y=revenue_by_age_group.values, palette="coolwarm")
plt.title("Revenue by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Revenue ($)")
plt.tight_layout()
plt.show()

# Return Impact Print
print(f"Estimated Total Revenue Loss Due to Returns: ${total_return_loss:,.2f}")

```
